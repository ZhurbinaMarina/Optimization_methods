```python
import sympy as sp
```

```python

# 1-3. Исходные данные, переменные и множители
x, y = sp.symbols("x y", real=True)
lam = sp.symbols("lambda", real=True)

variables = [x, y]
multipliers = [lam]

# Целевая функция и ограничения (Вариант 9)
f = x * y
h = x**2 + y**2 - 2
constraints = [h]

# Допуск для численной проверки
epsilon = 1e-8


# 4-6. Лагранжиан и система уравнений
# Строим L = f + lambda * h
L = f + sum(a * c for a, c in zip(multipliers, constraints))

# Система стационарности: производные по переменным + само ограничение
equations = [sp.diff(L, v) for v in variables] + constraints

# --- Вывод постановки задачи ---
print("Целевая функция:", f)
print("Ограничения:", h, "= 0")
print("Функция Лагранжа:", L)
print("Уравнения стационарности:")
for i, eq in enumerate(equations, 1):
    print(f"  {i}) {eq} = 0")
print()


# 7-12. Решение системы и отбор кандидатов
unknowns = variables + multipliers
# Решаем систему уравнений
solutions = sp.solve(equations, unknowns, dict=True)

valid_candidates = [] # Сюда сохраним "хорошие" точки

for number, solution in enumerate(solutions, start=1):
    # Если хотя бы одна переменная в решении содержит мнимую единицу (I), пропускаем корень
    if any(val.has(sp.I) for val in solution.values()):
        continue

    # Словарь только с координатами точки (без лямбды)
    point_subs = {v: solution[v] for v in variables}

    # Считаем невязку (подставляем точку в ограничение h)
    # sp.simplify() упростит математическое выражение до числа
    residual = sp.simplify(h.subs(point_subs))

    # Проверяем допустимость (модуль невязки меньше эпсилон)
    is_admissible = abs(float(sp.N(residual))) <= epsilon

    # Считаем значение целевой функции в этой точке
    value = sp.simplify(f.subs(point_subs))

    # Вытаскиваем значения для вывода
    pt_coords = tuple(solution[v] for v in variables)
    lam_val = solution[lam]

    # --- Вывод кандидата ---
    print(f"Кандидат {number}:")
    print(f"  точка: {pt_coords}")
    print(f"  множители: {lam_val}")
    print(f"  значение функции: {value}")
    print(f"  невязки ограничений: {residual}")
    print(f"  допустимость: {'да' if is_admissible else 'нет'}\n")

    if is_admissible:
        valid_candidates.append({'value': value})

print()
print("Итог:")

# 13. Классификация
if valid_candidates:
    min_val = min(c['value'] for c in valid_candidates)
    max_val = max(c['value'] for c in valid_candidates)

    print(f"  глобальный минимум: {min_val}")
    print(f"  глобальный максимум: {max_val}")
    print("  основание классификации: допустимое множество (окружность) компактно, целевая функция непрерывна. В силу регулярности задачи, экстремумы лежат среди стационарных точек. Сравнение значений функции доказывает глобальность.")
else:
    print("  Кандидаты не найдены.")
```
